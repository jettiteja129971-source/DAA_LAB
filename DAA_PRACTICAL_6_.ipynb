```python
# Implementation of Matrix Chain Multiplication
# Using Dynamic Programming

# Dimensions of matrices
# A1 = 10 x 20
# A2 = 20 x 30
# A3 = 30 x 40
# A4 = 40 x 30

p = [10, 20, 30, 40, 30]

# Number of matrices
n = len(p) - 1

# Create DP table
dp = [[0 for _ in range(n + 1)]
      for _ in range(n + 1)]

# Matrix chain multiplication
for length in range(2, n + 1):

    for i in range(1, n - length + 2):

        j = i + length - 1

        # Set initial cost to infinity
        dp[i][j] = float('inf')

        # Try all possible positions to split
        for k in range(i, j):

            cost = (dp[i][k] +
                    dp[k + 1][j] +
                    p[i - 1] * p[k] * p[j])

            # Store minimum multiplication cost
            if cost < dp[i][j]:
                dp[i][j] = cost

# Display DP table
print("Dynamic Programming Table:")

for i in range(1, n + 1):
    for j in range(1, n + 1):
        if i <= j:
            print(dp[i][j], end="\t")
        else:
            print("-", end="\t")
    print()

# Display minimum number of multiplications
print("\nMinimum number of scalar multiplications =",
      dp[1][n])
```
